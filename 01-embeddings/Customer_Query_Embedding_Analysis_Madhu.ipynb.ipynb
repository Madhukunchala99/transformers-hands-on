## Step 2: Subword Tokenization Using BERT

### 2.1 Loading the Tokenizer

We will use BERT's pretrained WordPiece tokenizer to split customer queries into words or smaller subword units.

The tokenizer uses an existing vocabulary, so we do not need to train it from scratch.

```python

%pip install -q transformers

from transformers import AutoTokenizer

model_name = "bert-base-uncased"

tokenizer = AutoTokenizer.from_pretrained(model_name)

print("Tokenizer Loaded Successfully!")
print("Vocabulary Size:", tokenizer.vocab_size)
```

### Observation

We successfully loaded BERT's pretrained tokenizer.

It contains 30,522 vocabulary tokens and uses WordPiece tokenization.

We will now use it to tokenize queries from our dataset.

### 2.2 Applying Subword Tokenization

We will select one customer query from our dataset and use BERT's WordPiece tokenizer to split it into tokens.

This helps us understand how BERT processes text before generating embeddings.

```python

text = df.loc[0, "customer_query"]

tokens = tokenizer.tokenize(text)

print("Original Sentence:", text)
print("Subword Tokens:", tokens)
print("Total Tokens:", len(tokens))
```

### Observation

BERT split the sentence into 8 tokens.

All the words were present in BERT's vocabulary, so no word required further subword splitting.

The tokenizer also converted the text into lowercase and treated the question mark as a separate token.

At this stage, we have generated tokens, not embeddings.


### 2.3 Converting Tokens into Token IDs

We will convert the subword tokens into numerical IDs using BERT's vocabulary.

BERT also adds two special tokens:
- [CLS] — Marks the beginning of the sequence.
- [SEP] — Marks the end of the sequence.

These IDs will later be used to retrieve embedding vectors.

```python

token_ids = tokenizer.encode(text)

tokens_with_special = tokenizer.convert_ids_to_tokens(token_ids)

print("Original Sentence:", text)
print("Tokens:", tokens_with_special)
print("Token IDs:", token_ids)
print("Total Tokens:", len(token_ids))
```

### Observation

The original sentence contained 8 tokens.

After adding [CLS] and [SEP], the total increased to 10.

Each token was mapped to its corresponding integer ID.

Token IDs are not embeddings; they are used to look up embedding vectors in the model.

### 2.4 Applying BERT Tokenization to the Complete Dataset

We will apply BERT's WordPiece tokenizer to all 300 customer queries.

For each query, we will generate subword tokens, token IDs, and token counts to understand how different sentences are processed.

```python

df["tokens"] = df["customer_query"].apply(
    lambda x: tokenizer.tokenize(x)
)

df["token_ids"] = df["customer_query"].apply(
    lambda x: tokenizer.encode(x)
)

df["token_count"] = df["token_ids"].apply(len)

print("Total Queries:", len(df))
print("Average Token Count:", round(df["token_count"].mean(), 2))
print("Maximum Token Count:", df["token_count"].max())

display(df.head(10))
```

## Step 3: Word Embeddings Using Word2Vec

### 3.1 Preparing Text for Word2Vec

Word2Vec converts individual words into numerical vectors based on their surrounding words.

Unlike BERT, Word2Vec uses word-level tokens rather than WordPiece subword tokens.

We will tokenize our 300 customer queries using NLTK, convert the words to lowercase, and remove punctuation before training Word2Vec.

```python

from nltk.tokenize import wordpunct_tokenize

sentences = [
    [
        word.lower()
        for word in wordpunct_tokenize(query)
        if word.isalpha()
    ]
    for query in df["customer_query"]
]

print("Total Sentences:", len(sentences))
print("First Sentence:", sentences[0])
print("Second Sentence:", sentences[1])
print("Third Sentence:", sentences[2])
```

### 3.2 Training Word2Vec

We will train a Word2Vec model using the Skip-gram approach, which learns word representations by predicting surrounding words.

**Parameters:**
- `vector_size=50`: Each word gets a 50-dimensional vector.
- `window=3`: Considers up to 3 nearby words on either side.
- `min_count=1`: Includes all words.
- `sg=1`: Uses Skip-gram.
- `epochs=200`: Trains for 200 passes over the dataset.

Our goal is to examine how words are represented as numerical vectors.

```python

%pip install -q gensim

from gensim.models import Word2Vec

word2vec_model = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=3,
    min_count=1,
    sg=1,
    epochs=200,
    seed=42,
    workers=1
)

print("Word2Vec Model Trained Successfully!")
print("Vocabulary Size:", len(word2vec_model.wv))
print("Embedding Dimension:", word2vec_model.vector_size)

word = "bank"
vector = word2vec_model.wv[word]

print("\nSelected Word:", word)
print("Word Vector:", vector)
print("Vector Shape:", vector.shape)
```

### Observation

We trained Word2Vec using 300 customer queries.

The model generates a fixed 50-dimensional vector for each word in its vocabulary.

These vectors capture patterns learned from surrounding words. However, the dataset is still small, so the quality of semantic relationships may be limited.

### 3.3 Measuring Word Similarity

We will calculate cosine similarity between different word embeddings.

Cosine similarity measures how closely two vectors point in the same direction, with values ranging from -1 to +1.

We will compare banking-related and unrelated words to examine whether Word2Vec has learned meaningful word relationships from our 300 queries.

```python

words_to_compare = [
    ("bank", "account"),
    ("bank", "loan"),
    ("bank", "river"),
    ("bank", "password"),
    ("card", "payment"),
    ("order", "refund")
]

print("Word Similarity Results:\n")

for word1, word2 in words_to_compare:
    similarity = word2vec_model.wv.similarity(word1, word2)

    print(f"{word1} vs {word2}: {similarity:.4f}")
```

### 3.4 Word2Vec Similarity Analysis

We calculated cosine similarity between different word embeddings trained on 300 customer queries.

**Key Observations:**

- Bank and account: 0.3639
- Bank and loan: 0.5158
- Bank and river: 0.5687
- Bank and password: -0.0615
- Card and payment: 0.2955
- Order and refund: 0.3390

The model identified different levels of similarity between words based on their surrounding context in the training data.

However, "bank" and "river" received a higher similarity score than "bank" and "account".

This shows that Word2Vec learns relationships from word usage patterns rather than directly understanding meaning.

Since Word2Vec generates a fixed vector for each word, it cannot distinguish the different meanings of "bank" in different sentences.

### 3.5 Finding the Most Similar Words

We will use Word2Vec's `most_similar()` function to find the five words whose embeddings are closest to "bank".

This helps us examine the relationships learned from our 300 customer queries.

```python

similar_words = word2vec_model.wv.most_similar(
    "bank",
    topn=5
)

print("Top 5 Most Similar Words to 'bank':\n")

for word, score in similar_words:
    print(f"{word}: {score:.4f}")
```

### 3.6 Final Observations — Word2Vec

We successfully trained Word2Vec using 300 customer queries.

**Results:**
- Vocabulary size: 439 words.
- Embedding dimension: 50.
- Each word has a fixed numerical vector.

The model identified "we" (0.6435), "along" (0.6409), and "walked" (0.6306) among the words most similar to "bank".

Some relationships were unexpected because our dataset is small and contains different meanings of the word "bank".

**Conclusion:**

Word2Vec helped us understand how individual words are represented as numerical vectors and compared using cosine similarity.

However, it uses a fixed embedding for each word and cannot directly distinguish different meanings of the same word based on sentence context.

## Step 4: Sentence Embeddings

### 4.1 Generating Sentence Embeddings

Word2Vec represents individual words as vectors, while sentence embeddings represent complete sentences as numerical vectors.

We will use the pretrained `all-MiniLM-L6-v2` model to generate embeddings for our 300 customer queries.

**Why are we doing this?**

To identify customer queries with similar meanings, even when they use different words.

```python

%pip install -q sentence-transformers

from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

queries = df["customer_query"].tolist()

sentence_embeddings = model.encode(
    queries,
    convert_to_tensor=True,
    show_progress_bar=True
)

print("Total Queries:", len(queries))
print("Embedding Shape:", sentence_embeddings.shape)
print("First Query:", queries[0])
print("First Embedding (10 values):", sentence_embeddings[0][:10])
```

### Observation

We generated sentence embeddings for 300 customer queries.

Each complete query is represented by a 384-dimensional vector.

The shape `(300, 384)` represents 300 queries, each containing 384 numerical features.

These embeddings can be used to calculate semantic similarity between complete sentences.

### 4.2 Calculating Sentence Similarity

We will calculate cosine similarity between selected customer queries using their sentence embeddings.

Unlike Word2Vec, which compares individual words, sentence embeddings allow us to compare the meanings of complete sentences.

Our goal is to check whether queries with similar meanings receive higher similarity scores than unrelated queries.

```python

from sentence_transformers import util

query_pairs = [
    (0, 1),      # Bank account queries
    (60, 61),    # Debit card queries
    (160, 161),  # Shopping queries
    (0, 280),    # Financial bank vs river bank
    (220, 221)   # Refund queries
]

for i, j in query_pairs:
    score = util.cos_sim(
        sentence_embeddings[i],
        sentence_embeddings[j]
    ).item()

    print("Query 1:", queries[i])
    print("Query 2:", queries[j])
    print(f"Similarity Score: {score:.4f}")
    print("-" * 60)
```

### 4.2 Observations

We calculated cosine similarity between selected customer queries using Sentence Transformer embeddings.

Related queries, such as opening a bank account and creating a savings account, received a similarity score of 0.6909.

The financial-bank and river-bank queries received a lower score of 0.2585, indicating different sentence meanings.

This demonstrates how sentence embeddings can capture semantic relationships between complete queries, even when their wording differs.

### 4.3 Finding the Most Similar Query

We will select one customer query and compare its embedding with all other queries in our dataset.

Using cosine similarity, we will identify the query with the highest similarity score.

This demonstrates how embeddings can support semantic search and customer query matching.

```python

import torch
from sentence_transformers import util

query_index = 0

scores = util.cos_sim(
    sentence_embeddings[query_index],
    sentence_embeddings
)[0].clone()

# Exclude the selected query itself
scores[query_index] = -float("inf")

best_match = torch.argmax(scores).item()

print("Customer Query:", queries[query_index])
print("Most Similar Query:", queries[best_match])
print("Similarity Score:", round(scores[best_match].item(), 4))
```

### 4.3 Observation

We compared one customer query against the remaining 299 queries using cosine similarity.

The model identified "I want to create a savings account" as the most similar query to "How can I open a bank account?", with a similarity score of 0.6909.

This demonstrates how sentence embeddings can identify related customer queries based on their meaning rather than exact word matching.

### 4.4 Finding the Top 5 Similar Queries

We will extend our semantic search to retrieve the five most similar customer queries.

This helps us understand how sentence embeddings can rank multiple queries according to semantic similarity.

```python

top_scores, top_indices = torch.topk(scores, k=5)

print("Original Query:", queries[query_index])
print("\nTop 5 Similar Customer Queries:\n")

for rank, (idx, score) in enumerate(
    zip(top_indices.tolist(), top_scores.tolist()),
    start=1
):
    print(f"{rank}. {queries[idx]}")
    print(f"   Similarity Score: {score:.4f}\n")
```

### 4.5 Final Observations — Sentence Embeddings

We generated 384-dimensional sentence embeddings for 300 customer queries using a pretrained Sentence Transformer.

Using cosine similarity, we identified the five closest matches for a selected customer query.

The most similar query was "I want to create a savings account", with a similarity score of 0.6909.

The model successfully retrieved semantically related queries, although some results had different customer intentions.

**Conclusion:** Sentence embeddings are useful for semantic search and identifying related queries, but high similarity does not always mean identical intent.

## Step 5: Contextual Embeddings Using BERT

### 5.1 Loading the BERT Model

Unlike Word2Vec, which assigns a fixed vector to each word, BERT generates contextual embeddings based on the surrounding words.

For example, "bank" in "river bank" and "bank account" has different meanings.

We will use the pretrained `bert-base-uncased` model to explore how BERT represents words in different contexts.

```python

import torch
from transformers import AutoTokenizer, AutoModel

model_name = "bert-base-uncased"

bert_tokenizer = AutoTokenizer.from_pretrained(model_name)
bert_model = AutoModel.from_pretrained(model_name)

bert_model.eval()

print("BERT Model Loaded Successfully!")
print("Hidden Size:", bert_model.config.hidden_size)
print("Number of Encoder Layers:", bert_model.config.num_hidden_layers)
```

### 5.2 Generating Contextual Embeddings

We will pass two sentences containing the word "bank" through BERT.

- "I sat near the river bank."
- "The bank approved my home loan."

BERT processes each token along with its surrounding words and generates contextual embeddings.

We will examine the resulting token representations and their dimensions.

```python

sentences_bert = [
    "I sat near the river bank.",
    "The bank approved my home loan."
]

inputs = bert_tokenizer(
    sentences_bert,
    return_tensors="pt",
    padding=True,
    truncation=True
)

with torch.no_grad():
    outputs = bert_model(**inputs)

contextual_embeddings = outputs.last_hidden_state

print("Input Sentences:")
for sentence in sentences_bert:
    print(sentence)

print("\nInput IDs Shape:", inputs["input_ids"].shape)
print("Contextual Embeddings Shape:", contextual_embeddings.shape)
print("Embedding Dimension:", contextual_embeddings.shape[-1])
```

### Observation

BERT generated contextual embeddings for both sentences.

The embedding shape `(2, 9, 768)` represents:

- 2 input sentences.
- 9 token positions per sentence, including special tokens and padding.
- 768 numerical values per token.

These token representations are generated by BERT's encoder and incorporate information from the surrounding context.

### 5.3 Comparing the Word "bank" in Different Contexts

We will extract the embedding of "bank" from both sentences and compare them using cosine similarity.

Although the word is identical, its meaning changes depending on the surrounding words.

Our objective is to examine whether BERT generates different contextual representations for the same word.

```python

import torch.nn.functional as F

# Find the position of "bank" in each sentence
tokens_1 = bert_tokenizer.convert_ids_to_tokens(
    inputs["input_ids"][0].tolist()
)

tokens_2 = bert_tokenizer.convert_ids_to_tokens(
    inputs["input_ids"][1].tolist()
)

bank_index_1 = tokens_1.index("bank")
bank_index_2 = tokens_2.index("bank")

# Extract contextual embeddings
bank_vector_1 = contextual_embeddings[0, bank_index_1]
bank_vector_2 = contextual_embeddings[1, bank_index_2]

# Calculate cosine similarity
similarity = F.cosine_similarity(
    bank_vector_1.unsqueeze(0),
    bank_vector_2.unsqueeze(0)
).item()

print("Sentence 1:", sentences_bert[0])
print("Bank Token Position:", bank_index_1)
print("Embedding Shape:", bank_vector_1.shape)

print("\nSentence 2:", sentences_bert[1])
print("Bank Token Position:", bank_index_2)
print("Embedding Shape:", bank_vector_2.shape)

print("\nCosine Similarity:", round(similarity, 4))
print("Embeddings Identical:", torch.allclose(
    bank_vector_1, bank_vector_2
))
```

### 5.4 Final Observations — Contextual Embeddings

We generated contextual embeddings using pretrained BERT.

The word "bank" was examined in two different sentences representing river-related and financial contexts.

Both embeddings contained 768 dimensions, but their values were different.

The cosine similarity between them was 0.4509, and the embeddings were not identical.

**Conclusion:** Unlike Word2Vec, BERT generates contextual representations based on surrounding words and token positions, helping represent different meanings of the same word.

### 5.5 Comparing the Word "charge" in Different Contexts

We will examine the word "charge" in two sentences:

- "Please charge my mobile phone." — Charging a battery.
- "The bank applied an additional service charge." — A financial fee.

We will extract BERT embeddings for "charge" and compare them using cosine similarity to observe how context affects its representation.

```python

sentences_charge = [
    "Please charge my mobile phone.",
    "The bank applied an additional service charge."
]

charge_inputs = bert_tokenizer(
    sentences_charge,
    return_tensors="pt",
    padding=True,
    truncation=True
)

with torch.no_grad():
    charge_outputs = bert_model(**charge_inputs)

charge_embeddings = charge_outputs.last_hidden_state

for i, sentence in enumerate(sentences_charge):
    tokens = bert_tokenizer.convert_ids_to_tokens(
        charge_inputs["input_ids"][i].tolist()
    )

    print(f"Sentence {i+1}:", sentence)
    print("Tokens:", tokens)

    charge_index = tokens.index("charge")

    if i == 0:
        charge_vector_1 = charge_embeddings[i, charge_index]
    else:
        charge_vector_2 = charge_embeddings[i, charge_index]

similarity_charge = torch.nn.functional.cosine_similarity(
    charge_vector_1.unsqueeze(0),
    charge_vector_2.unsqueeze(0)
).item()

print("\nEmbedding Shape:", charge_vector_1.shape)
print("Cosine Similarity:", round(similarity_charge, 4))
print("Embeddings Identical:", torch.allclose(
    charge_vector_1, charge_vector_2
))
```

### 5.7 Contextual Embeddings — Different Meanings of "Interest"

We will compare the word "interest" in two sentences:

- "The bank offers interest on savings accounts."
- "My interest in technology is growing."

The first sentence refers to financial interest, while the second refers to personal curiosity.

We will extract BERT's contextual embeddings and calculate cosine similarity to examine how the representations differ.

```python

sentences_interest = [
    "The bank offers interest on savings accounts.",
    "My interest in technology is growing."
]

interest_inputs = bert_tokenizer(
    sentences_interest,
    return_tensors="pt",
    padding=True,
    truncation=True
)

with torch.no_grad():
    interest_outputs = bert_model(**interest_inputs)

interest_embeddings = interest_outputs.last_hidden_state

interest_vectors = []

for i, sentence in enumerate(sentences_interest):
    tokens = bert_tokenizer.convert_ids_to_tokens(
        interest_inputs["input_ids"][i].tolist()
    )

    interest_index = tokens.index("interest")
    vector = interest_embeddings[i, interest_index]

    interest_vectors.append(vector)

    print(f"Sentence {i+1}:", sentence)
    print("Token Position:", interest_index)
    print("Embedding Shape:", vector.shape)
    print("First 5 Values:", vector[:5])
    print()

similarity_interest = torch.nn.functional.cosine_similarity(
    interest_vectors[0].unsqueeze(0),
    interest_vectors[1].unsqueeze(0)
).item()

print("Cosine Similarity:", round(similarity_interest, 4))
print("Embeddings Identical:", torch.allclose(
    interest_vectors[0],
    interest_vectors[1]
))
```

### 5.7 Observation

We compared the word "interest" in financial and personal-interest contexts.

Both embeddings contained 768 dimensions, but their numerical values differed.

The cosine similarity was 0.4348, showing that the two contextual representations were not identical.

This demonstrates how BERT represents the same word differently depending on sentence context and position.

### 5.8 Comparing the Same Word in Similar Contexts

We will compare the word "bank" in two sentences:

- "The bank approved my home loan."
- "The bank approved my personal loan."

Both sentences use "bank" as a financial institution.

We will calculate cosine similarity to examine whether BERT produces closely aligned embeddings when the meaning and token position remain similar.

```python

sentences_same_context = [
    "The bank approved my home loan.",
    "The bank approved my personal loan."
]

same_inputs = bert_tokenizer(
    sentences_same_context,
    return_tensors="pt",
    padding=True,
    truncation=True
)

with torch.no_grad():
    same_outputs = bert_model(**same_inputs)

same_embeddings = same_outputs.last_hidden_state

bank_vectors = []

for i, sentence in enumerate(sentences_same_context):
    tokens = bert_tokenizer.convert_ids_to_tokens(
        same_inputs["input_ids"][i].tolist()
    )

    bank_index = tokens.index("bank")
    vector = same_embeddings[i, bank_index]

    bank_vectors.append(vector)

    print(f"Sentence {i+1}:", sentence)
    print("Token Position:", bank_index)
    print("Embedding Shape:", vector.shape)
    print()

similarity_same = torch.nn.functional.cosine_similarity(
    bank_vectors[0].unsqueeze(0),
    bank_vectors[1].unsqueeze(0)
).item()

print("Cosine Similarity:", round(similarity_same, 4))
print("Embeddings Identical:", torch.allclose(
    bank_vectors[0], bank_vectors[1]
))
```

BERT generated 768-dimensional embeddings for each token.

The same word received different numerical representations depending on its sentence context.

In our experiment, similar financial contexts produced highly similar embeddings, while different meanings produced lower similarity scores.

**Conclusion:** Unlike Word2Vec, which generates one fixed vector per word, BERT generates contextual embeddings influenced by surrounding words and token positions.

## Step 6: Positional Encoding

### 6.1 Understanding Word Positions

Embeddings represent tokens as numerical vectors, but Transformer models also need information about token order.

Consider two sentences:

- "The customer paid the merchant."
- "The merchant paid the customer."

Both contain the same words, but their meanings differ because of word order.

**Why positional encoding?**

Positional encoding adds information about each token's position to its embedding, allowing Transformer layers to use word-order information.

We will demonstrate sinusoidal positional encoding separately using PyTorch. BERT itself uses learned positional embeddings.

### 6.2 Tokenizing Sentences with Different Word Orders

We will tokenize two sentences containing the same words in different orders.

Our goal is to identify the position of each token before generating positional encodings.

```python

position_sentences = [
    "The customer paid the merchant.",
    "The merchant paid the customer."
]

for i, sentence in enumerate(position_sentences):
    tokens = bert_tokenizer.tokenize(sentence)

    print(f"\nSentence {i+1}: {sentence}")

    for position, token in enumerate(tokens):
        print(f"Position {position}: {token}")
```

### 6.3 Generating Positional Encodings

We will use sine and cosine functions to generate a numerical vector for each token position.

Each position receives a fixed encoding based on its location in the sentence.

We will use 8 dimensions initially to make the output easy to understand.

**Why?** Positional encodings provide position information that Transformer models can combine with token embeddings.

```python

import torch
import math

seq_len = 6
d_model = 8

positions = torch.arange(seq_len).unsqueeze(1)
dimensions = torch.arange(0, d_model, 2)

div_term = torch.exp(
    dimensions * (-math.log(10000.0) / d_model)
)

positional_encoding = torch.zeros(seq_len, d_model)

positional_encoding[:, 0::2] = torch.sin(
    positions * div_term
)

positional_encoding[:, 1::2] = torch.cos(
    positions * div_term
)

print("Positional Encoding Shape:", positional_encoding.shape)
print("\nPositional Encoding Matrix:")
print(positional_encoding.round(decimals=4))
```

### 6.4 Combining Token Embeddings with Positional Encodings

We will create 8-dimensional token embeddings for our two sentences and add the positional encodings generated earlier.

**Formula:**

Input Representation = Token Embedding + Positional Encoding

We will examine how the same word receives different input representations when its position changes.

This is a simplified Transformer input demonstration, not BERT's actual positional embedding implementation.

```python

import torch

torch.manual_seed(42)

# Tokenize both sentences
tokens_1 = bert_tokenizer.tokenize(position_sentences[0])
tokens_2 = bert_tokenizer.tokenize(position_sentences[1])

# Create a shared embedding for each unique token
vocab = sorted(set(tokens_1 + tokens_2))

embedding_layer = torch.nn.Embedding(
    num_embeddings=len(vocab),
    embedding_dim=d_model
)

token_to_id = {token: i for i, token in enumerate(vocab)}

ids_1 = torch.tensor([token_to_id[t] for t in tokens_1])
ids_2 = torch.tensor([token_to_id[t] for t in tokens_2])

# Generate token embeddings
embeddings_1 = embedding_layer(ids_1)
embeddings_2 = embedding_layer(ids_2)

# Add positional encodings
final_1 = embeddings_1 + positional_encoding
final_2 = embeddings_2 + positional_encoding

# Compare the word "customer"
index_1 = tokens_1.index("customer")
index_2 = tokens_2.index("customer")

print("Word: customer")
print("Position in Sentence 1:", index_1)
print("Position in Sentence 2:", index_2)

print("\nOriginal Embeddings Identical:",
      torch.allclose(embeddings_1[index_1],
                     embeddings_2[index_2]))

print("\nFinal Embedding Sentence 1:")
print(final_1[index_1].detach().round(decimals=4))

print("\nFinal Embedding Sentence 2:")
print(final_2[index_2].detach().round(decimals=4))

print("\nFinal Embeddings Identical:",
      torch.allclose(final_1[index_1],
                     final_2[index_2]))
```

### 6.5 Final Observations — Positional Encoding

We implemented sinusoidal positional encoding using PyTorch.

We compared two sentences containing the same words in different orders.

The word "customer" appeared at position 1 in the first sentence and position 4 in the second.

Before adding positional encoding, its token embeddings were identical.

After adding positional encoding, the final representations became different.

**Conclusion:** Positional encoding provides word-order information to Transformer models by combining position vectors with token embeddings.

This experiment demonstrates positional encoding using randomly initialized token embeddings. Actual BERT models use learned positional embeddings.

### 6.6 Comparing Positional Encodings for "merchant"

We will examine the word "merchant" in the same two sentences.

- Sentence 1: "The customer paid the merchant."
- Sentence 2: "The merchant paid the customer."

We will compare its original token embeddings and final representations after adding positional encodings.

Our goal is to verify how changing a word's position affects its numerical representation.

```python

# Find merchant positions
merchant_index_1 = tokens_1.index("merchant")
merchant_index_2 = tokens_2.index("merchant")

# Original token embeddings
merchant_original_1 = embeddings_1[merchant_index_1]
merchant_original_2 = embeddings_2[merchant_index_2]

# Embeddings after adding positional encoding
merchant_final_1 = final_1[merchant_index_1]
merchant_final_2 = final_2[merchant_index_2]

print("Word: merchant")
print("Position in Sentence 1:", merchant_index_1)
print("Position in Sentence 2:", merchant_index_2)

print("\nOriginal Embeddings Identical:",
      torch.allclose(merchant_original_1, merchant_original_2))

print("\nFinal Embedding Sentence 1:")
print(merchant_final_1.detach().round(decimals=4))

print("\nFinal Embedding Sentence 2:")
print(merchant_final_2.detach().round(decimals=4))

print("\nFinal Embeddings Identical:",
      torch.allclose(merchant_final_1, merchant_final_2))
```

### 6.6 Observation

We compared the word "merchant" at positions 4 and 1 in two sentences.

The original token embeddings were identical because both sentences used the same embedding lookup.

After adding sinusoidal positional encodings, the final representations became different.

This confirms that positional encoding introduces token-position information into the input representations.

Both positional encoding experiments produced the expected results.
